This appendix lab represents a stripped-down block header as a Python dataclass and prints a human-readable summary. It connects the field list from the text (version, prev hash, Merkle root, time, bits, nonce) to a concrete object you can inspect or adapt to real data sources.

```python
from dataclasses import dataclass  # structured data


@dataclass
class Header:  # minimal header model
    version: int
    prev_hash: str
    merkle_root: str
    time: int  # Unix-style timestamp
    bits: int
    nonce: int


hdr = Header(  # toy header values
    version=0x20000000,
    prev_hash="000000...prev",
    merkle_root="4d5e6f...root",
    time=1_700_000_000,
    bits=0x170fffff,
    nonce=42,
)  # end of header

print("Block header summary:")
print("  version   :", hex(hdr.version))
print("  prev_hash :", hdr.prev_hash)
print("  merkle    :", hdr.merkle_root)
print("  time      :", hdr.time)
print("  bits      :", hex(hdr.bits))
print("  nonce     :", hdr.nonce)
```
