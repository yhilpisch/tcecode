## Appendix B – Difficulty and Hash-Rate Calculations

This appendix snippet turns the difficulty and hash-rate formulas into code. You compute an expected block time for a given network hash rate and target success probability, and then estimate an attacker’s share of total hash power for back-of-the-envelope reorg risk checks.

```python
import math  # access to math functions


def expected_block_time(hash_rate: float, target_prob: float) -> float:
    """Return expected seconds per block for hash_rate and success probability."""
    return 1.0 / (hash_rate * target_prob)


def attacker_share(attacker_hash: float, network_hash: float) -> float:
    """Return an attacker's fraction of total hash power."""
    return attacker_hash / network_hash


network_hash = 300e15  # 300 PH/s (toy value)
target_prob = 2**-32  # toy success probability per hash

et = expected_block_time(network_hash, target_prob)  # seconds per block
print("Expected block time (min):", et / 60)

attacker = 30e15  # 30 PH/s attacker
print("Attacker share:", attacker_share(attacker, network_hash))
```
