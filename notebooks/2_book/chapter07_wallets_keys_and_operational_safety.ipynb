Here you explore a toy hierarchical key derivation scheme inspired by HD wallets. A single root seed and a human-readable path deterministically produce child "keys," giving you a tangible way to think about account branches and change chains without touching real funds or full standards.

```python
import hashlib  # hashing primitives
import hmac  # keyed hashing
from typing import Tuple  # type annotations (kept for clarity)


def derive_child(root: bytes, path: str) -> bytes:
    """Deterministically derive a child key from a root and a path.

    This is a toy example inspired by HD wallet ideas and is
    not a drop-in replacement for real standards.
    """
    msg = path.encode("utf-8")  # encode path as bytes
    return hmac.new(root, msg, hashlib.sha256).digest()  # keyed hash


def format_key(b: bytes) -> str:
    """Return a short hex representation for display."""
    return b.hex()[:16]  # first 8 bytes as 16 hex chars


seed = b"demo-seed-entropy"  # toy root secret (bytes)

paths = [  # example derivation paths
    "m/personal/receive/0",
    "m/personal/receive/1",
    "m/personal/change/0",
    "m/business/receive/0",
]

for p in paths:  # iterate over example paths
    child_key = derive_child(seed, p)  # derive key
    print(p, "->", format_key(child_key))  # show truncated hex
```
