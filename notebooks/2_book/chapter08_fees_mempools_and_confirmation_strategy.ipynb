This notebook cell implements a tiny mempool and block template builder. By sorting transactions by fee rate and filling a limited block, you see how miners optimise revenue per vByte rather than absolute fees, tying code directly to the chapter’s fee-market discussion.

```python
from dataclasses import dataclass  # structured records
from typing import List  # type hints


@dataclass
class Tx:  # simplified transaction model
    txid: str  # identifier
    size_vb: int  # size in vBytes
    fee_sat: int  # fee in satoshis

    @property
    def fee_rate(self) -> float:
        """Return fee rate in sat/vByte."""
        return self.fee_sat / self.size_vb


def select_for_block(mempool: List[Tx], max_vb: int) -> List[Tx]:
    """Select transactions by descending fee rate up to a size limit."""
    chosen: List[Tx] = []
    used = 0  # total vBytes used so far
    for tx in sorted(mempool, key=lambda t: t.fee_rate, reverse=True):
        if used + tx.size_vb <= max_vb:  # fits into block?
            chosen.append(tx)
            used += tx.size_vb
    return chosen


mempool = [  # toy mempool snapshot
    Tx("tx1", size_vb=200, fee_sat=4_000),   # 20 sat/vB
    Tx("tx2", size_vb=300, fee_sat=12_000),  # 40 sat/vB
    Tx("tx3", size_vb=150, fee_sat=1_500),   # 10 sat/vB
    Tx("tx4", size_vb=500, fee_sat=25_000),  # 50 sat/vB
]

block_limit_vb = 700  # toy block size limit in vBytes

chosen = select_for_block(mempool, block_limit_vb)  # build template

print("Selected transactions in order:")
for tx in chosen:
    print(tx.txid, "fee_rate=", f"{tx.fee_rate:.1f}", "sat/vB")
```
