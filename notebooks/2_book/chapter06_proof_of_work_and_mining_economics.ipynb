You start by turning the PoW success probability into an expected block time. Given a hash rate and per-hash success probability, this function computes how long, on average, you would wait to find a block—linking the abstract target parameter to concrete seconds and minutes.

```python
def expected_block_time(hash_rate: float, success_prob: float) -> float:
    """Return expected seconds to find a block."""
    # 1 / (trials per second * success probability)
    return 1.0 / (hash_rate * success_prob)


hash_rate = 1e12  # 1 terahash per second (TH/s)
target_ratio = 2**-32  # toy success probability per hash (not real difficulty)

et = expected_block_time(hash_rate, target_ratio)  # expected seconds
print("Expected time (s):", et)
print("Expected time (min):", et / 60)
```

Next, you simulate the randomness of block discovery using exponential inter-arrival times. This reinforces the idea that "10 minutes" is an average over many blocks, not a deterministic schedule, and shows how variable real block intervals can be.

```python
import random  # pseudo-random number generator
import math  # logarithms for exponential sampling


def sample_inter_block_time(lam: float) -> float:
    """Sample one exponential inter-block time with rate lam (blocks per second)."""
    u = random.random()  # uniform in [0, 1)
    return -math.log(1.0 - u) / lam  # inverse CDF for exponential


network_hash_rate = 1e15  # 1 petahash per second (PH/s)
success_prob = 2**-32  # same toy probability per hash as above
lam = network_hash_rate * success_prob  # expected blocks per second

# draw 10 inter-block times
samples = [sample_inter_block_time(lam) for _ in range(10)]

print("Inter-block times (min):")
for t in samples:
    print(f"  {t/60:.2f}")
```

Finally, you sketch a toy mining revenue calculator. It takes a miner’s hash-rate share, the current block reward, and a BTC/USD price to estimate expected daily revenue, connecting protocol parameters back to basic economic incentives.

```python
block_reward_btc = 6.25  # BTC per block (example)
btc_price_usd = 30_000  # USD per BTC (example)

miner_hash_rate = 100e12  # 100 TH/s
network_hash_rate = 200e15  # 200 PH/s

blocks_per_day = 144  # approximate (24h * 60 / 10)

share = miner_hash_rate / network_hash_rate  # miner's fraction of hash power
expected_blocks = share * blocks_per_day  # expected blocks per day

revenue_usd = expected_blocks * block_reward_btc * btc_price_usd  # daily revenue

print("Expected blocks/day:", expected_blocks)
print("Expected revenue/day (USD):", revenue_usd)
```
