You begin by deriving a key from a password using PBKDF2. This mirrors real-world setups where humans remember a passphrase but systems need a fixed-length key with tunable work factors to slow down offline guessing attacks.

```python
import hashlib
import os

password = "correct horse battery staple".encode("utf-8")
salt = os.urandom(16)  # random salt per user/record
iterations = 200_000  # work factor; adjust as hardware evolves

dk = hashlib.pbkdf2_hmac("sha256", password, salt, iterations, dklen=32)
print("Derived key (hex):", dk.hex())
print("Salt (hex):", salt.hex())
print("Iterations:", iterations)
```

Next, you wrap this derivation in a tiny storage format. The idea is to bundle salt, iteration count, and derived key in a single string so that verification later uses the exact same parameters, making password checks reproducible and explicit.

```python
import hashlib
import os

def derive_password_record(password: str) -> str:
    salt = os.urandom(16)
    iterations = 150_000
    dk = hashlib.pbkdf2_hmac(
        "sha256", password.encode("utf-8"), salt, iterations, dklen=32
    )
    return f"pbkdf2_sha256${iterations}${salt.hex()}${dk.hex()}"

record = derive_password_record("s3cret")
print("Stored record:", record)
```

Finally, you draw a truly random key and nonce from the operating system. This exercise reinforces that key material and nonces should come from a cryptographically strong source of randomness, not from ad hoc generators or predictable sequences.

```python
import os

key = os.urandom(32)  # 256-bit symmetric key
nonce = os.urandom(12)  # 96-bit nonce often used with AEAD

print("Key (hex):", key.hex())
print("Nonce (hex):", nonce.hex())
```
