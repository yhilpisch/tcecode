This toy example implements the simplest possible "stream cipher": XORing plaintext with a keystream. It is deliberately insecure, but it gives you a concrete mental model for how real stream ciphers combine a secret key, a nonce, and a pseudorandom stream to turn readable bytes into ciphertext.

```python
def xor_stream(data: bytes, key: bytes) -> bytes:  # toy XOR stream cipher
    """XOR each plaintext byte with the corresponding key byte.

    This is not secure, but it makes the keystream idea tangible.
    """
    return bytes(d ^ k for d, k in zip(data, key))

plaintext = b"attack at dawn"  # message to "encrypt"
keystream = b"supersecretstream"[: len(plaintext)]  # truncate to match length

ciphertext = xor_stream(plaintext, keystream)  # encrypt
print("Ciphertext bytes:", ciphertext)

recovered = xor_stream(ciphertext, keystream)  # decrypt by XORing again
print("Recovered plaintext:", recovered)
```

Next, you demonstrate the classic keystream-reuse failure. By encrypting two different messages with the same XOR keystream, you reveal structure in the XOR of ciphertexts—exactly why real systems must never reuse a (key, nonce) pair.

```python
def xor_bytes(a: bytes, b: bytes) -> bytes:
    """XOR two byte strings up to the length of the shorter one."""
    return bytes(x ^ y for x, y in zip(a, b))

msg1 = b"pay alice 10"  # first plaintext
msg2 = b"pay bob 10  "  # second plaintext (note padding)

ks = b"samekeystream!!!"[: max(len(msg1), len(msg2))]  # reused keystream

c1 = xor_stream(msg1, ks)  # encrypt first message
c2 = xor_stream(msg2, ks)  # encrypt second message with same keystream

xor_ct = xor_bytes(c1, c2)  # XOR of ciphertexts
print("c1 XOR c2:", xor_ct)
print("(Should equal msg1 XOR msg2):", xor_bytes(msg1, msg2))
```
