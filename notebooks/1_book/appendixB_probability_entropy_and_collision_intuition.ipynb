You start by estimating passphrase entropy under a simple uniform model. The goal is not to endorse real-world passwords, but to practice turning alphabet size and length into approximate bits so you can reason about brute-force cost.

```python
import math

alphabet_size = 26  # e.g. lowercase letters
length = 10  # characters in the passphrase

entropy_bits = length * math.log2(alphabet_size)
print("Alphabet size:", alphabet_size)
print("Length:", length)
print("Idealised entropy (bits):", entropy_bits)
```

Next you simulate collisions for randomly drawn nonces. By counting how many draws it takes before you hit a repeat, you build intuition for the birthday-bound curves discussed in the appendix.

```python
import secrets  # cryptographically strong randomness

def simulate_collisions(trials: int, bits: int) -> int:
    """Draw nonces until a collision occurs or we hit the trial limit.

    Returns the number of draws actually performed.
    """
    seen = set()
    for i in range(1, trials + 1):
        nonce = secrets.token_bytes(bits // 8)
        if nonce in seen:
            return i  # collision at draw i
        seen.add(nonce)
    return trials  # no collision within the trial budget

for bits in (16, 24):  # toy sizes so collisions are observable
    draws = simulate_collisions(10_000, bits)
    print(f"bits={bits}, draws until collision or cap={draws}")
```

Finally, you connect "security bits" to rough work estimates. The appendix uses this line of reasoning whenever it translates bit levels into more intuitive statements about how much brute force an attacker would need on average.

```python
for bits in (32, 64, 80, 128):
    space = 2**bits
    avg_work = space / 2  # average brute-force trials
    print(f"{bits:3d} bits -> space≈2^{bits}, average work≈{avg_work:.2e} trials")
```
