This first mini-lab lets you poke at modular inverses numerically. You can change the modulus and base to see when an inverse exists, and verify that the product really lands back on 1 modulo `n`, connecting the abstract extended Euclidean algorithm to concrete numbers.

```python
def mod_inverse(a: int, n: int) -> int | None:  # return a^{-1} mod n if it exists
    """Compute the modular inverse of a modulo n using a brute-force search.

    This is for intuition only; later you will use more efficient algorithms.
    """
    a = a % n  # normalise a into 0..n-1
    for x in range(1, n):  # try all possible candidates
        if (a * x) % n == 1:  # check inverse condition
            return x
    return None  # no inverse exists if we never hit 1

pairs = [(3, 11), (2, 8), (7, 26)]  # (a, n) examples to explore
for a, n in pairs:  # iterate over (a, n) pairs
    inv = mod_inverse(a, n)  # attempt to find inverse
    if inv is None:
        print(f"a={a} has no inverse modulo n={n}")
    else:
        print(f"a={a}, n={n} -> inverse={inv}, check={(a * inv) % n}")
```

Next, you translate the birthday-bound intuition into a tiny function you can experiment with. By changing the sample count `k` and the space size `N`, you see how quickly collision probability grows and why even modest key or nonce sizes must be chosen with care.

```python
import math  # exponential function for the birthday bound

def birthday_collision_prob(k: int, N: int) -> float:
    """Approximate collision probability for k samples from a space of size N."""
    if k <= 1:
        return 0.0
    return 1 - math.exp(-k * (k - 1) / (2 * N))  # standard birthday-bound form

N = 10**6  # toy space size (e.g., 20-bit space)
for k in (10, 100, 500, 1000):  # sample sizes to compare
    p = birthday_collision_prob(k, N)  # compute approximate collision probability
    print(f"k={k:4d}, N={N:>8d} -> collision probability ≈ {p:.6f}")
```
