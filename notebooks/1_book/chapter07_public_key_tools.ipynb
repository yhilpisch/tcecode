You begin with a toy Diffie–Hellman-style key exchange in a tiny modular arithmetic setting. Even though the parameters are insecure, the code shows how two parties can arrive at the same shared secret while only sending public values over the wire.

```python
p = 23  # small prime modulus (insecure but easy to inspect)
g = 5   # generator modulo p

a = 6  # Alice's secret exponent
b = 15 # Bob's secret exponent

A = pow(g, a, p)  # Alice's public value
B = pow(g, b, p)  # Bob's public value

shared_alice = pow(B, a, p)  # Alice computes shared secret
shared_bob = pow(A, b, p)    # Bob computes shared secret

print("A (Alice -> Bob):", A)
print("B (Bob -> Alice):", B)
print("Shared (Alice):", shared_alice)
print("Shared (Bob):  ", shared_bob)
print("Match?", shared_alice == shared_bob)
```

Next, you sketch the "hash-before-sign" pattern: rather than signing raw, structured data directly, you first compute a digest and then sign that fixed-size hash. This is a structural practice exercise only, using placeholders where a real signature primitive would go.

```python
import hashlib

message = b"release manifest v1.2"  # structured data to sign
digest = hashlib.sha256(message).digest()  # fixed-size hash

# In a real system, you would call a signature API here, passing `digest`.
# For this lab we just display the digest to emphasise the separation.
print("Message:", message)
print("SHA-256 digest (hex):", digest.hex())
```
