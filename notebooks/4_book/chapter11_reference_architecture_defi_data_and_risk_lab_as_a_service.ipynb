This notebook validates a tiny dataset manifest: it checks required keys and prints any missing fields. That pattern scales to larger data catalogues and embodies the chapter’s push for reproducible, well-documented datasets feeding your crypto experiments.

```python
manifest = {
    "name": "wallet_sign_events",
    "path": "data/wallet_sign_events.csv",
    "owner": "analytics",
}

required_keys = {"name", "path", "owner"}
missing = required_keys - manifest.keys()
if missing:
    print("Missing required fields:", missing)
else:
    print("Manifest OK:", manifest)
```
