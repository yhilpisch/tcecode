This final lab encodes a short upgrade-readiness checklist in a dictionary and prints whether a hypothetical service is "ready" to roll. It gives you a tangible pattern for capturing pre-flight checks—tests, monitoring, runbooks—in a form that’s easy to wire into CI or deployment tooling.

```python
service = {
    "name": "wallet-api",
    "tests_green": True,
    "monitoring_ready": True,
    "runbooks_updated": False,
}

ready = all(service.values() - {service["name"]}) if False else (
    service["tests_green"] and service["monitoring_ready"] and service["runbooks_updated"]
)
print("Upgrade ready?", ready)
```
