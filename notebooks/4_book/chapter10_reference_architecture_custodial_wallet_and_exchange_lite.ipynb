This helper builds a small manifest of controls, metrics, and runbooks in memory and prints it in a readable way. It shows how to keep the relationships between controls and evidence explicit, just as the chapter’s "control/monitoring manifest" template suggests.

```python
manifest = {
    "controls": [
        {"id": "c_sign_latency", "metric": "wallet.sign.latency",
         "runbook": "runbooks/sign_latency.md"},
        {"id": "c_queue_depth", "metric": "wallet.queue.depth",
         "runbook": "runbooks/queue_depth.md"},
    ]
}

for c in manifest["controls"]:
    print(f"Control {c['id']}: metric={c['metric']}, runbook={c['runbook']}")
```
