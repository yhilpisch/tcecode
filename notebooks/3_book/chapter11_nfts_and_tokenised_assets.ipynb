This example inspects a tiny piece of NFT-style metadata stored as JSON. By looking at fields like `name`, `description`, and `image`, you connect the chapter’s discussion of on-chain vs. off-chain content to a concrete document you can tweak and extend.

```python
import json
from pathlib import Path

meta_path = Path("toy_nft_metadata.json")

if not meta_path.exists():
    sample_meta = {
        "name": "Example NFT #1",
        "description": "Tiny demo metadata for Cryptomarkets and Systems.",
        "image": "ipfs://examplehash",
        "attributes": [{"trait_type": "rarity", "value": "common"}],
    }
    meta_path.write_text(json.dumps(sample_meta, indent=2), encoding="utf-8")

meta = json.loads(meta_path.read_text(encoding="utf-8"))
print("Name:", meta.get("name"))
print("Image URI:", meta.get("image"))
print("Attributes:", meta.get("attributes"))
```
