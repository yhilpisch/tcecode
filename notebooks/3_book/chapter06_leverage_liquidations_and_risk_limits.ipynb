This cell explores a toy liquidation threshold: given collateral, debt, and a maintenance margin, you compute the critical price at which a leveraged position is at risk. It turns the chapter’s formulas into a small, tweakable calculator you can play with directly in a notebook.

```python
collateral_value = 10_000.0  # current USD value of collateral
debt = 5_000.0  # borrowed amount
maintenance_margin = 1.3  # collateral/debt ratio required

# Critical collateral value where collateral/debt == maintenance_margin
critical_collateral = maintenance_margin * debt
drawdown_needed = 1 - critical_collateral / collateral_value

print("Critical collateral value:", critical_collateral)
print("Drawdown until risk (fraction):", round(drawdown_needed, 4))
```
