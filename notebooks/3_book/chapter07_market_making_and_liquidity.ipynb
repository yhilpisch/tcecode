This toy simulation sketches a symmetric market maker posting bid and ask quotes around a mid-price. It tracks inventory and cash as random buy/sell hits arrive, illustrating how spread revenue and inventory risk trade off over time without invoking any external data or libraries.

```python
import random

mid = 100.0
spread = 1.0
bid = mid - spread / 2
ask = mid + spread / 2

inventory = 0.0
cash = 0.0

for _ in range(20):  # 20 random hits
    side = random.choice(["buy", "sell"])  # from the taker's perspective
    size = 1.0
    if side == "buy":  # taker buys from us at ask
        inventory -= size
        cash += size * ask
    else:  # taker sells to us at bid
        inventory += size
        cash -= size * bid

pnl = cash + inventory * mid
print("Final inventory:", inventory)
print("Final cash:", cash)
print("Approx P&L at mid:", pnl)
```
