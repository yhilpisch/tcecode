Here you inspect a tiny JSON snapshot of a lending market—deposits, borrows, and reserves. This mirrors the chapter’s practice of freezing protocol state into compact records you can diff, chart, and reason about without connecting to live infrastructure.

```python
import json
from pathlib import Path

snapshot_path = Path("toy_lending_snapshot.json")

if not snapshot_path.exists():
    sample = {
        "total_deposits": 1_000_000.0,
        "total_borrows": 600_000.0,
        "reserves": 50_000.0,
        "base_rate": 0.02,
        "slope": 0.1,
    }
    snapshot_path.write_text(json.dumps(sample, indent=2), encoding="utf-8")

snap = json.loads(snapshot_path.read_text(encoding="utf-8"))

utilisation = snap["total_borrows"] / snap["total_deposits"]
print("Total deposits:", snap["total_deposits"])
print("Total borrows:", snap["total_borrows"])
print("Utilisation:", round(utilisation, 3))
```
