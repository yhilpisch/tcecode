This lab simulates a miniature CI gate: given a list of recent tests and their outcomes, it decides whether to allow a crypto-sensitive change to merge. It highlights how to encode “must-pass” suites as data rather than tribal knowledge.

```python
tests = [
    {"name": "test_aead_roundtrip", "status": "pass"},
    {"name": "test_kdf_parameters", "status": "pass"},
    {"name": "test_randomness_source", "status": "pass"},
]

required = {"test_aead_roundtrip", "test_kdf_parameters"}
passed = {t["name"] for t in tests if t["status"] == "pass"}

missing = required - passed
if missing:
    print("CI gate FAILED; missing:", missing)
else:
    print("CI gate PASSED; all required crypto tests green.")
```
