This helper walks a toy key inventory and performs a few simple sanity checks: missing owners, stale rotation dates, and unclear storage locations. It mirrors the chapter’s recommendation to keep key inventories explicit and machine-readable so that audits and reviews can flag drift early.

```python
from datetime import date

keys = [
    {"id": "signing_key_hot", "owner": "security", "last_rotated": "2024-01-10", "location": "hsm_cluster_a"},
    {"id": "api_token_analytics", "owner": "data", "last_rotated": "2023-06-01", "location": "env_var"},
]

today = date.today()

def days_since(d: str) -> int:
    y, m, d_ = map(int, d.split("-"))
    return (today - date(y, m, d_)).days

print("Keys needing attention:")
for k in keys:
    age = days_since(k["last_rotated"])
    if age > 365:
        print("  -", k["id"], "rotation overdue (days:", age, ")")
```
