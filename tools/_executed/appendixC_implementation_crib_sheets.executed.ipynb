The first snippet demonstrates explicit encoding and decoding between text and bytes. This reinforces the habit of handling encodings deliberately rather than relying on implicit conversions that can break across systems.

```python
text = "café"  # includes a non-ASCII character
data = text.encode("utf-8")  # encode to bytes explicitly
roundtrip = data.decode("utf-8")  # decode back to text

print("Text:", text)
print("Bytes:", data)
print("Round-trip:", roundtrip)
```

Next you use `hmac.compare_digest` to perform constant-time tag comparisons. This tiny habit reduces the risk of subtle timing side channels when checking MACs or password hashes.

```python
import hmac
import hashlib

key = b"mac-key"
msg = b"important message"

tag = hmac.new(key, msg, hashlib.sha256).digest()
candidate = hmac.new(key, msg, hashlib.sha256).digest()

print("Tags equal?", hmac.compare_digest(tag, candidate))
```

Finally, you draw nonces and tokens from the `secrets` module. This mirrors the appendix’s guidance to rely on vetted randomness APIs for identifiers, tokens, and one-time values instead of home-grown generators.

```python
import secrets

nonce = secrets.token_bytes(12)  # 96-bit nonce for AEAD-like schemes
token = secrets.token_urlsafe(32)  # URL-safe random token for identifiers

print("Nonce (hex):", nonce.hex())
print("Token:", token)
```
