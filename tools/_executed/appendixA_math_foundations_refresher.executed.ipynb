This first snippet checks, by brute force, which residues have multiplicative inverses modulo a fixed `n`. It turns the textbook rule “inverse exists iff `gcd(a, n) == 1`” into a quick experiment you can run whenever you want to sanity-check arithmetic over small moduli.

```python
import math  # access to gcd

def has_inverse(a: int, n: int) -> bool:
    """Return True if a has a multiplicative inverse modulo n."""
    return math.gcd(a, n) == 1  # inverse exists iff gcd(a, n) == 1

for a in range(1, 13):  # sample values for a
    print(f"a={a}, has inverse mod 12? {has_inverse(a, 12)}")
```

Next you approximate the birthday bound numerically. By plugging different sample counts `k` into the same function, you see how quickly collision probability rises even for modest spaces, reinforcing the intuition behind nonce and hash-length choices.

```python
import math  # access to exp for approximations

def birthday_prob(k: int, N: int) -> float:
    """Approximate collision probability for k draws from N possibilities."""
    return 1 - math.exp(-k * (k - 1) / (2 * N))

N = 2**16  # sample space size (like a 16-bit nonce)
for k in [100, 500, 2000]:
    p = birthday_prob(k, N)
    print(f"k={k}, N={N}, p≈{p:.4f}")
```

Finally, you connect raw counts to "bits of entropy" via `log2`. This is the mental bridge the appendix uses whenever it turns large keyspaces or password dictionaries into more intuitive bit-sized security levels.

```python
import math  # mathematical functions including log2

options = 10_000_000  # number of possibilities
bits = math.log2(options)  # bits of entropy for uniform choices

print("Options:", options)
print("Entropy (bits):", bits)
```
