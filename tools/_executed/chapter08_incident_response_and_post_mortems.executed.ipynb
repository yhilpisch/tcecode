This helper appends an incident record to a JSON log file under `runbooks/`. It gives you a concrete pattern for capturing severity, timestamps, owners, and control IDs in a structured way, supporting the chapter’s emphasis on evidence-driven incident reviews.

```python
import json
from datetime import datetime
import pathlib

INCIDENT_LOG = pathlib.Path("runbooks/incidents.json")
INCIDENT_LOG.parent.mkdir(parents=True, exist_ok=True)

record = {
    "id": "INC-2025-04",
    "severity": "S1",
    "opened_at": datetime.utcnow().isoformat(),
    "status": "open",
    "summary": "Signer latency above SLO, withdrawals paused",
    "owners": {
        "commander": "alex",
        "comms": "morgan",
        "scribe": "lee",
    },
    "controls": ["c_sign_latency", "c_queue_depth"],
    "mttd_seconds": 420,
    "mttr_seconds": None,
}

log = []
if INCIDENT_LOG.exists():
    log = json.loads(INCIDENT_LOG.read_text())
log.append(record)
INCIDENT_LOG.write_text(json.dumps(log, indent=2))
print(f"Recorded incident {record['id']}")
```
