This cell inspects a simplified smart-contract transaction record stored in JSON. By loading fields such as `gas_limit`, `gas_used`, and `gas_price`, you can compute total fee paid and reason about how gas mechanics turn abstract execution into concrete costs.

```python
import json
from pathlib import Path

tx_path = Path("toy_smart_tx.json")

if not tx_path.exists():
    sample_tx = {
        "from": "0xabc...",
        "to": "0xdef...",
        "gas_limit": 100_000,
        "gas_used": 80_000,
        "gas_price_gwei": 30,
    }
    tx_path.write_text(json.dumps(sample_tx, indent=2), encoding="utf-8")

tx = json.loads(tx_path.read_text(encoding="utf-8"))

gas_used = tx["gas_used"]
gas_price_gwei = tx["gas_price_gwei"]
fee_eth = gas_used * gas_price_gwei / 1e9  # 1 gwei = 1e-9 ETH

print("Gas used:", gas_used)
print("Gas price (gwei):", gas_price_gwei)
print("Fee (ETH):", fee_eth)
```
