This lab encodes a minimal list of services and the trust boundaries between them, then prints where sensitive assets cross contexts. It turns the chapter’s call for explicit boundary mapping into a concrete data structure you can diff and extend with your own systems.

```python
services = [
    {"name": "api_gateway", "trust_zone": "edge", "assets": ["access_tokens"]},
    {"name": "wallet_backends", "trust_zone": "core", "assets": ["signing_keys"]},
    {"name": "analytics", "trust_zone": "analytics", "assets": ["aggregated_metrics"]},
]

boundaries = [
    ("edge", "core"),
    ("core", "analytics"),
]

print("Trust boundaries:")
for src, dst in boundaries:
    print(f"  {src} -> {dst}")
```
