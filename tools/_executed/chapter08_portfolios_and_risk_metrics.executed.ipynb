This lab works with a toy table of asset returns and a vector of portfolio weights. You compute portfolio returns and a basic volatility estimate, echoing the chapter’s emphasis on building an intuition for diversification and risk before reaching for heavier quantitative toolkits.

```python
# rows: time, columns: assets A, B
returns_A = [0.01, -0.02, 0.015, 0.0]
returns_B = [0.005, 0.01, -0.005, 0.02]
weights = [0.6, 0.4]

portfolio_returns = []
for ra, rb in zip(returns_A, returns_B):
    rp = weights[0] * ra + weights[1] * rb
    portfolio_returns.append(rp)

mean_rp = sum(portfolio_returns) / len(portfolio_returns)
var_rp = sum((r - mean_rp) ** 2 for r in portfolio_returns) / (len(portfolio_returns) - 1)
vol_p = var_rp ** 0.5

print("Portfolio returns:", [round(r, 4) for r in portfolio_returns])
print("Mean return:", round(mean_rp, 4))
print("Volatility:", round(vol_p, 4))
```
