```python
risk_memo = {
    "system": "example_cex_integration",
    "scenario": "exchange withdrawal freeze",
    "checks": [
        "max exposure per venue defined",
        "read-only fallback mode documented",
        "proof-of-reserves usage understood",
    ],
}

for k, v in risk_memo.items():
    print(f"{k.upper()}:", v)
```
