We start by turning the narrative idea of a "chain of blocks" into an explicit Python structure. This tiny example models three block headers as dictionaries linked via `prev_hash` fields so you can see how each block points back to its predecessor, mirroring the way real Bitcoin nodes verify hash links when they walk the best chain.

```python
blocks = [  # toy chain of three blocks
    {
        "height": 0,  # genesis block height
        "hash": "0000000...genesis",  # placeholder hash
        "prev_hash": None,  # no previous block
    },
    {
        "height": 1,  # next block in the chain
        "hash": "0000001...block1",  # placeholder hash
        "prev_hash": "0000000...genesis",  # points to genesis hash
    },
    {
        "height": 2,  # third block
        "hash": "0000002...block2",  # placeholder hash
        "prev_hash": "0000001...block1",  # points to prior block
    },
]  # end of toy chain

for b in blocks:  # iterate over blocks
    print(  # show height and hash link
        f"height={b['height']}, hash={b['hash']}, prev={b['prev_hash']}"
    )
```

Next, you encode the classic "Alice pays Bob" story as a simple threat model. Instead of keeping this in prose only, you capture assets, adversaries, assumptions, and controls in a dictionary that can live in version control. This makes it easy to diff how your understanding of Bitcoin risk evolves as you read later chapters and build real tools.

```python
threat_model = {  # structured view of one payment
    "assets": [  # what we want to protect
        "Alice's spendable coins",
        "Bob's receipt of 0.01 BTC",
        "ledger integrity for this payment",
    ],
    "adversaries": [  # who might attack
        "double-spender broadcasting conflicts",
        "miner coalition attempting a short reorg",
        "network attacker eclipsing a node",
        "malicious or buggy wallet software",
    ],
    "assumptions": [  # conditions we rely on
        "no single miner has majority hash power",
        "Alice and Bob see multiple peers",
        "wallet implementations are well-audited",
    ],
    "controls": [  # mitigations in and around Bitcoin
        "wait for sufficient confirmations",
        "use full-node verification",
        "cross-check chain tips from several sources",
    ],
}  # end of threat model

for k, v in threat_model.items():  # iterate for display
    print(f"{k.upper()}:")  # section heading
    for item in v:  # each element in section
        print("  -", item)  # bullet-like line
```
