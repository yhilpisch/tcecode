Here you replicate a command-line checksum inside Python. This lets you verify downloads or configuration files programmatically, using the same SHA-256 hashes that tools like `shasum` produce.

```python
import hashlib
from pathlib import Path

path = Path("example.bin")
if not path.exists():
    path.write_bytes(b"demo data for checksum")  # create a small demo file

h = hashlib.sha256()
with path.open("rb") as f:
    for chunk in iter(lambda: f.read(4096), b""):
        h.update(chunk)

print("SHA-256:", h.hexdigest())
```
