This notebook cell works with a tiny CSV of block-level metrics—heights, timestamps, transaction counts, and fees. By loading the file and computing a few simple aggregates, you mirror the chapter’s message that many on-chain insights start from small, well-structured tables rather than heavyweight analytics stacks.

```python
import csv  # CSV parsing
from pathlib import Path  # path handling

blocks_path = Path("blocks_sample.csv")

if not blocks_path.exists():  # create a small sample dataset
    sample_rows = [
        {"height": 1, "tx_count": 120, "total_fees": 0.05, "avg_fee_per_tx": 0.0004},
        {"height": 2, "tx_count": 90, "total_fees": 0.03, "avg_fee_per_tx": 0.0003},
        {"height": 3, "tx_count": 150, "total_fees": 0.08, "avg_fee_per_tx": 0.0005},
    ]
    fieldnames = ["height", "tx_count", "total_fees", "avg_fee_per_tx"]
    with blocks_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(sample_rows)

tx_counts = []
total_fees = []

with blocks_path.open(newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        tx_counts.append(int(row["tx_count"]))
        total_fees.append(float(row["total_fees"]))

print("Blocks loaded:", len(tx_counts))
print("Average tx_count:", sum(tx_counts) / len(tx_counts))
print("Total fees across sample:", sum(total_fees))
```
