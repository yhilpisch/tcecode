This helper finds a “metric aggregator” configuration in a small in-memory registry. It reflects the chapter’s encouragement to keep metric definitions and ownership centralised so you can audit which dashboards and alerts depend on which signals.

```python
metrics = [
    {"name": "wallet.sign.latency", "owner": "wallet_team", "agg": "p95"},
    {"name": "exchange.depth.btcusd", "owner": "markets", "agg": "min"},
]

target = "wallet.sign.latency"
for m in metrics:
    if m["name"] == target:
        print("Aggregator for", target, "is", m["agg"])
        break
```
