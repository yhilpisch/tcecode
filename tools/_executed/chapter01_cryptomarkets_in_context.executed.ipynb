This notebook turns the chapter’s "market atlas" warm-up into runnable code. Instead of pulling live data, you work with a tiny, local CSV of hand-curated cryptomarket events so that everything runs deterministically in Colab. The goal is to practice loading structured event data and preview how sectors and narratives show up over time without touching keys, APIs, or real money.

```python
import csv  # built-in CSV reader
from pathlib import Path  # convenient path handling

# In the book, the code assumes an existing CSV in a data directory.
# For a self-contained Colab notebook, we create a tiny sample file on the fly
# if it is not present yet.

events_path = Path("crypto_events_sample.csv")  # local file in the notebook workspace

if not events_path.exists():  # create a small sample dataset
    sample_events = [
        {"date": "2017-12-17", "sector": "Bitcoin", "label": "Bitcoin peaks near 20k USD"},
        {"date": "2020-03-12", "sector": "DeFi", "label": "DeFi stress-test during market crash"},
        {"date": "2020-12-01", "sector": "Ethereum", "label": "Ethereum 2.0 beacon chain launches"},
        {"date": "2021-05-19", "sector": "Market", "label": "Leverage flush and broad crypto sell-off"},
        {"date": "2022-11-08", "sector": "CEX", "label": "Major centralised exchange collapse"},
    ]

    fieldnames = ["date", "sector", "label"]
    with events_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(sample_events)

events = []  # list of event dicts
with events_path.open(newline="", encoding="utf-8") as f:  # open CSV file
    reader = csv.DictReader(f)  # read rows as dicts
    for row in reader:  # iterate over rows
        events.append(row)  # collect event

print(f"Loaded {len(events)} events")  # quick sanity check
for e in events[:3]:  # show first few
    print(e["date"], e["sector"], "->", e["label"])  # basic preview
```
